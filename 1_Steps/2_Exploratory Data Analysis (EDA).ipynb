```python
# Exploratory Data Analysis (EDA)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

theme = sns.set_theme(style="whitegrid")
df = pd.read_csv(r'C:\Users\Soumya_SRB\Desktop\L&T Project Intelligencee\India_Construction_Data_L&T.csv')
print("Number of projects:", df.shape[0])
print("Number of columns:", df.shape[1])
display(df.head())

# How many projects are labelled as high delay risk vs not high delay risk.
risk_counts = df["Delay_Risk"].value_counts()
print(risk_counts)

risk_counts.plot(kind="bar",
                 figsize=(7, 4),
                 title="Construction project delay risk")

plt.xlabel("Risk of delay")
plt.ylabel("Number of Projects")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df,
            x="Project_Progress_Percent",
            bins=20,
            kde=True)

plt.title("Distribution of Project Progress")
plt.xlabel("Project Progress (%)")
plt.ylabel("Number of Projects >")
plt.tight_layout()
plt.show()
```

```python
## Material Availability vs Delay Risk.
plt.figure(figsize=(7, 4))

sns.countplot(
    data=df,
    x="Material_Availability",
    hue="Delay_Risk",
    order=["Low", "Medium", "High"])

plt.title("Material Availability vs Delay Risk")
plt.xlabel("Material Availability")
plt.ylabel("Number of Projects")
plt.tight_layout()
plt.show()

## Equipment Availability vs Delay Risk.
plt.figure(figsize=(7, 4))

sns.countplot(
    data=df,
    x="Equipment_Availability",
    hue="Delay_Risk",
    order=["Low", "Medium", "High"])

plt.title("Equipment Availability vs Delay Risk")
plt.xlabel("Equipment Availability")
plt.ylabel("Number of Projects")
plt.tight_layout()
plt.show()
```

```python
## Compare weather risk with delay risk.
plt.figure(figsize=(7, 4))

sns.countplot(
    data=df,
    x="Weather_Risk",
    hue="Delay_Risk",
    order=["Low", "Medium", "High"])

plt.title("Weather Risk vs Project Delay Risk")
plt.xlabel("Weather Risk")
plt.ylabel("Number of Projects")
plt.tight_layout()
plt.show()
```

```python
## Examine project complexity.

plt.figure(figsize=(7, 4))

sns.countplot(
    data=df,
    x="Project_Complexity",
    hue="Delay_Risk",
    order=["Low", "Medium", "High"])

plt.title("Project Complexity vs Delay Risk")
plt.xlabel("Project Complexity")
plt.ylabel("Number of Projects")
plt.tight_layout()
plt.show()
```

```python
## Correlation heatmap.
numeric_df = df.select_dtypes(include="number").drop(
    columns=["Project_ID"],
    errors="ignore")

plt.figure(figsize=(11, 7))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5)

plt.title("Correlation Between Numerical Project Features")
plt.tight_layout()
plt.show()
```

```python
## Compare numerical features by delay risk.

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Delay_Risk",
    y="Project_Progress_Percent",
    hue="Delay_Risk")

plt.title("Project Progress by Delay Risk Category")
plt.xlabel("Delay Risk")
plt.ylabel("Project Progress (%)")
plt.tight_layout()
plt.show()
```

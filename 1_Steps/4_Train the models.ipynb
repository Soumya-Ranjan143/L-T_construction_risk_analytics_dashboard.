```python
from pathlib import Path
import joblib

# Correct path to the Outputs folder.
OUTPUT_DIR = Path(r"C:\Users\Soumya_SRB\Desktop\L&T Project Intelligencee\Outputs")

# Check that the saved file exists
prepared_file = OUTPUT_DIR /"prepared_data.joblib"

if not prepared_file.exists():
    raise FileNotFoundError(f"File not found: {prepared_file}\n"
        "Run the save cell at the end of Notebook 3 first.")

# Load the saved data.
data = joblib.load(prepared_file)

X_train = data["X_train"]
X_test = data["X_test"]
y_train = data["y_train"]
y_test = data["y_test"]
preprocessor = data["preprocessor"]

print("Prepared data loaded successfully!")
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("Preprocessor loaded successfully.")
```

```python
## 4.Train the models.

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

# Baseline model
logistic_model = Pipeline(
    [("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000))])

# Random Forest model
forest_model = Pipeline(
    [("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced"))])

# Train the models
logistic_model.fit(X_train, y_train)
forest_model.fit(X_train, y_train)

print("Both models trained successfully.")
```

```python
# Predict delay risk for the test data
logistic_predictions = logistic_model.predict(X_test)
forest_predictions = forest_model.predict(X_test)

print("Logistic Regression predictions:")
print(logistic_predictions[:10])

print("\nRandom Forest predictions:")
print(forest_predictions[:10])
```

```python
from pathlib import Path
import joblib

Outputs = Path(r"C:\Users\Soumya_SRB\Desktop\L&T Project Intelligencee\Outputs")

joblib.dump(
    {"logistic_model": logistic_model,
     "forest_model": forest_model},
      Outputs / "trained_models.joblib")

print("Both trained models saved.")
```
